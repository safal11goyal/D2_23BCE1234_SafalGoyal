## Load Data


```python
import pandas as pd

df = pd.read_csv('/content/crop-yield.csv')

print("First 5 rows of the DataFrame:")
print(df.head())
```

## Prepare Data


```python
y = df['Crop_Yield_ton_per_hectare']
X = df.drop('Crop_Yield_ton_per_hectare', axis=1)

print(f"Shape of features (X): {X.shape}")
print(f"Shape of target (y): {y.shape}")
```

```python
print("Missing values in X:")
print(X.isnull().sum()[X.isnull().sum() > 0])
```

```python
X_encoded = pd.get_dummies(X, drop_first=True)

print(f"Shape of features after one-hot encoding (X_encoded): {X_encoded.shape}")
print("First 5 rows of X after one-hot encoding:")
print(X_encoded.head())
```

```python
from sklearn.model_selection import train_test_split

X_train, X_test, y_train, y_test = train_test_split(X_encoded, y, test_size=0.2, random_state=42)

print(f"Shape of X_train: {X_train.shape}")
print(f"Shape of X_test: {X_test.shape}")
print(f"Shape of y_train: {y_train.shape}")
print(f"Shape of y_test: {y_test.shape}")
```

## Train Linear Regression



```python
from sklearn.linear_model import LinearRegression
from sklearn.metrics import mean_absolute_error, mean_squared_error, r2_score

# 3. Instantiate a LinearRegression model
linear_reg_model = LinearRegression()

# 4. Train the model using the X_train and y_train datasets
linear_reg_model.fit(X_train, y_train)

# 5. Make predictions on the X_test dataset
y_pred_lr = linear_reg_model.predict(X_test)

# 6. Calculate the evaluation metrics
mae_lr = mean_absolute_error(y_test, y_pred_lr)
mse_lr = mean_squared_error(y_test, y_pred_lr)
r2_lr = r2_score(y_test, y_pred_lr)

# 7. Print the calculated MAE, MSE, and R2 score
print("Linear Regression Model Evaluation:")
print(f"Mean Absolute Error (MAE): {mae_lr:.4f}")
print(f"Mean Squared Error (MSE): {mse_lr:.4f}")
print(f"R-squared (R2) Score: {r2_lr:.4f}")
```

## Train Decision Tree



```python
from sklearn.tree import DecisionTreeRegressor

# Instantiate a DecisionTreeRegressor model
decision_tree_model = DecisionTreeRegressor(random_state=42)

# Train the model using the X_train and y_train datasets
decision_tree_model.fit(X_train, y_train)

# Make predictions on the X_test dataset
y_pred_dt = decision_tree_model.predict(X_test)

# Calculate the evaluation metrics
mae_dt = mean_absolute_error(y_test, y_pred_dt)
mse_dt = mean_squared_error(y_test, y_pred_dt)
r2_dt = r2_score(y_test, y_pred_dt)

# Print the calculated MAE, MSE, and R2 score
print("Decision Tree Regression Model Evaluation:")
print(f"Mean Absolute Error (MAE): {mae_dt:.4f}")
print(f"Mean Squared Error (MSE): {mse_dt:.4f}")
print(f"R-squared (R2) Score: {r2_dt:.4f}")
```

## Train Random Forest



```python
from sklearn.ensemble import RandomForestRegressor

# Instantiate a RandomForestRegressor model
random_forest_model = RandomForestRegressor(random_state=42)

# Train the model using the X_train and y_train datasets
random_forest_model.fit(X_train, y_train)

# Make predictions on the X_test dataset
y_pred_rf = random_forest_model.predict(X_test)

# Calculate the evaluation metrics
mae_rf = mean_absolute_error(y_test, y_pred_rf)
mse_rf = mean_squared_error(y_test, y_pred_rf)
r2_rf = r2_score(y_test, y_pred_rf)

# Print the calculated MAE, MSE, and R2 score
print("Random Forest Regression Model Evaluation:")
print(f"Mean Absolute Error (MAE): {mae_rf:.4f}")
print(f"Mean Squared Error (MSE): {mse_rf:.4f}")
print(f"R-squared (R2) Score: {r2_rf:.4f}")
```

## Train Gradient Boosting


```python
from sklearn.ensemble import GradientBoostingRegressor

# Instantiate a GradientBoostingRegressor model
gradient_boosting_model = GradientBoostingRegressor(random_state=42)

# Train the model using the X_train and y_train datasets
gradient_boosting_model.fit(X_train, y_train)

# Make predictions on the X_test dataset
y_pred_gbr = gradient_boosting_model.predict(X_test)

# Calculate the evaluation metrics
mae_gbr = mean_absolute_error(y_test, y_pred_gbr)
mse_gbr = mean_squared_error(y_test, y_pred_gbr)
r2_gbr = r2_score(y_test, y_pred_gbr)

# Print the calculated MAE, MSE, and R2 score
print("Gradient Boosting Regression Model Evaluation:")
print(f"Mean Absolute Error (MAE): {mae_gbr:.4f}")
print(f"Mean Squared Error (MSE): {mse_gbr:.4f}")
print(f"R-squared (R2) Score: {r2_gbr:.4f}")
```

## Compare Model Performance and Summarize Results





```python
import pandas as pd

# Create a dictionary to store the evaluation metrics
results = {
    'Model': ['Linear Regression','Decision Tree', 'Random Forest', 'Gradient Boosting'],
    'MAE': [mae_lr, mae_dt, mae_rf, mae_gbr],
    'MSE': [mse_lr, mse_dt, mse_rf, mse_gbr],
    'R2 Score': [r2_lr, r2_dt, r2_rf, r2_gbr]
}

# Create a DataFrame from the results dictionary
results_df = pd.DataFrame(results)

# Sort by R2 Score in descending order for better comparison
results_df = results_df.sort_values(by='R2 Score', ascending=False).reset_index(drop=True)

print("\nModel Performance Summary:\n")
print(results_df.to_markdown(index=False))
```
